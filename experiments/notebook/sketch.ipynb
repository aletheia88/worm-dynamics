```python
## visualize data
num_worms = data.shape[0]
num_variables = data.shape[1]

worm_idx = np.random.choice(list(range(num_worms)))
variables = {0: 'AVA', 1: 'MC', 2: 'SMDV', 3: 'velocity', 4: 'pumping', 5: 'head angle'}

fig, axs = plt.subplots(num_variables, 1, figsize=(12, 2 * num_variables), sharex=True, sharey=True)

for i in range(num_variables):

    axs[i].plot(data[worm_idx, i, :], label=f'{variables[i]}')
    axs[i].grid(True)
    axs[i].legend()
    axs[i].set_title(f'{variables[i]} from {datasets[worm_idx]}')
```
