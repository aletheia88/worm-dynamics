```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
dataset_path = "/home/alicia/notebook/alicia/attention_predict/data/AVA_MC_all.npy"
data = np.load(dataset_path)
```

```python
data.shape
```

```python
num_worms = data.shape[0]

fig, axs = plt.subplots(num_worms, 1, figsize=(12, 2 * num_worms), sharex=True, sharey=True)
idx = 1

for worm_idx in range(num_worms):

    axs[worm_idx].plot(data[worm_idx, idx, :], label=f'trace {worm_idx}')
    axs[worm_idx].grid(True)
    axs[worm_idx].legend()
    print(f'max activity: {np.max(data[worm_idx, idx, :])} min activity: {np.min(data[worm_idx, idx, :])}')
```
